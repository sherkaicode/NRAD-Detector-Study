```python
import os
import subprocess
```

```python
# --- Paths and Executables ---
DELPHES_EXE = "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/Delphes/DelphesHepMC2"
ROOT_DIR = "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/Root_Files"

# Input HEPMC files
# bsm_hepmc = "/home/aegis/Titan0/pythia2/pythia8245/examples/Research/hepmc_lxplus/1000_2_events.hepmc"
ttbar_file = "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/run_01/tag_1_pythia8_events.hepmc"

# Dictionary of Delphes cards
cards_dir = {
    "ATLAS_default": "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/Delphes Cards/delphes_card_ATLAS.tcl",
    "ATLAS_PileUp_default": "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/Delphes Cards/delphes_card_ATLAS_PileUp.tcl",
    "ATLAS_Run2": "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/Delphes Cards/delphes_card_ATLAS_Run2.tcl",
    "ATLAS_PileUp_Run2": "/home/aegis/Titan0/madgraph/MG5_aMC_v3_7_0/bin/ttbar_validation/Events/Delphes Cards/delphes_card_ATLAS_PileUp_Run2.tcl",
}

# Ensure output directory exists
os.makedirs(ROOT_DIR, exist_ok=True)
```

```python
phi_m = 1000

for r_inv in [4, 6]:
    print(f"\nProcessing phi_m={phi_m} GeV and r_inv=0.{r_inv}...")
    
    # Loop through each card and run Delphes for both BSM and ttbar events
    for card_name, card_path in cards_dir.items():
        print("-" * 60)
        print(f" - Processing Card: {card_name}")
        print(f"   Card File:   {card_path}")
        
        # Define output ROOT filenames
        bsm_output_root = os.path.join(ROOT_DIR, f"{phi_m}_{r_inv}_{card_name}_events.root")
        bsm_hepmc = f"/home/aegis/Titan0/pythia2/pythia8245/examples/Research/hepmc_lxplus/{phi_m}_{r_inv}_events.hepmc"
        # ttbar_output_root = os.path.join(ROOT_DIR, f"ttbar_{card_name}_events.root")
        
        print(f"   BSM Input HEPMC: {bsm_hepmc}")
        print(f"   BSM Output ROOT: {bsm_output_root}")
        
        # Run Delphes for BSM events
        subprocess.run([DELPHES_EXE, card_path, bsm_output_root, bsm_hepmc])
        print("   [SUCCESS] BSM Simulation completed.")
        
        # Run Delphes for ttbar events
        # (Output is likely suppressed here based on your notebook logs, but it runs successfully)
        # subprocess.run(
        #     [DELPHES_EXE, card_path, ttbar_output_root, ttbar_file], 
        #     stdout=subprocess.DEVNULL, # Suppresses standard output to keep terminal clean
        #     stderr=subprocess.DEVNULL  # Suppresses error output 
        # )
        # print("   [SUCCESS] ttbar Simulation completed.")
```
